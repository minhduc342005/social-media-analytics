```python
import pandas as pd
df = pd.read_excel('Social_Media_Content_Performance_Dataset.xlsx')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
df.columns.tolist()
```

```python
df['month'] = df['post_date'].dt.month
df['day_of_week'] = df['post_date'].dt.day_name()
df['quarter'] = df['post_date'].dt.quarter
df.head()
```

```python
df.columns.tolist()
```

```python
df.to_csv('social_media_cleaned.csv', index=False)
print("Cleaned dataset saved as 'social_media_cleaned.csv'")
```
